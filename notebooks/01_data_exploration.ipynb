**<h1>Real-Time Financial Fraud & Anomaly Detection Platform</h1>**

```python
import pandas as pd

df = pd.read_csv(
    "../data/transactions.csv",
    low_memory=False
)

print("Shape:", df.shape)

print("\nFraud:")
print(df["is_fraud"].value_counts(dropna=False))

print("\nUnique IDs:")

print("Users:", df["user_id"].nunique())
print("Merchants:", df["merchant_id"].nunique())
print("Devices:", df["device_id"].nunique())
print("IPs:", df["ip_address"].nunique())

print("\nDuplicate transaction IDs:")
print(df["transaction_id"].duplicated().sum())

print("\nMissing:")
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
#Checking for anomalies
print(
    "user == merchant:",
    (df["user_id"] == df["merchant_id"]).mean()
)

print(
    "user == device:",
    (df["user_id"] == df["device_id"]).mean()
)

print(
    "merchant == device:",
    (df["merchant_id"] == df["device_id"]).mean()
)
```
